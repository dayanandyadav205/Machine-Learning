```python
co_att = [2.58, 2.53, 2.59, 2.57, 2.73]
print("Simple avg of all COs:", sum(co_att)/len(co_att))

matrix = {
    'PO1': [3, 3, 3, 3, 2],
    'PO2': [2, 3, 2, 2, 1],
    'PO3': [0, 1, 3, 3, 0],
    'PO5': [0, 0, 2, 2, 0],
    'PSO1': [2, 2, 2, 2, 1],
    'PSO2': [0, 0, 2, 2, 0],
    'PSO3': [1, 2, 3, 3, 1],
}

for k, weights in matrix.items():
    num = sum(w * a for w, a in zip(weights, co_att))
    den = sum(weights)
    weighted_avg = num / den

    # Also check unweighted average of mapped COs only
    mapped_att = [a for w, a in zip(weights, co_att) if w > 0]
    unweighted_mapped = sum(mapped_att) / len(mapped_att)

    # Also check (Weight / 3) * CO_att averaged over mapped COs
    scaled_att = [(w / 3.0) * a for w, a in zip(weights, co_att) if w > 0]
    avg_scaled = sum(scaled_att) / len(scaled_att)

    print(f"{k}: Weighted={weighted_avg:.2f}, Unweighted_Mapped={unweighted_mapped:.2f}, Scaled_Avg={avg_scaled:.2f}")
```
