```python
import pandas as pd

# Read dataset
df = pd.read_csv("data.csv")

# 1. View the dataset
print("First 5 rows:\n", df.head(5))

# 2. Check for missing values
print("\nMissing values per column:\n", df.isnull().sum())

# 3. Handle missing values
# Option A: Fill missing numeric values with mean
df['Age'] = df['Age'].fillna(df['Age'].mean())

# Option B: Fill missing categorical values with mode
df['Country'] = df['Country'].fillna(df['Country'].mode()[0])

# Option C: Drop rows with too many missing values
df = df.dropna(thresh=3)   # keeps rows with at least 3 non-null values

# 4. Handle categorical features
# Convert categorical column into numeric codes
df['Country_Code'] = df['Country'].astype('category').cat.codes
df['Gender_Male'] = df['Gender'].astype('category').cat.codes
df['Country_India'] = df['Country'].astype('category').cat.codes
df['Country_USA'] = df['Country'].astype('category').cat.codes


# One-hot encoding for categorical features
df = pd.get_dummies(df, columns=['Gender', 'Country'], drop_first=True)

# 5. Verify changes
print("\nProcessed dataset:\n", df.head())
```
