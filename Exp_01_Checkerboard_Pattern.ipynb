```python
 #Imports the NumPy library with the alias 'np'.
import numpy as np 

# Defines a function named 'checkerboard' that takes one parameter 'N', which represents the size of the board.
def checkerboard(N): 
    # Creates an N×N array filled with zeros, using integer data type.
    board = np.zeros((N, N), dtype=int) 

    # Fill alternating positions with 1
    # Odd rows, even columns
    board[1::2, ::2]  = 1
    # Even rows, odd columns
    board[::2, 1::2]  = 1

    return board

# Example usage
N = 8
print(checkerboard(N))
```
