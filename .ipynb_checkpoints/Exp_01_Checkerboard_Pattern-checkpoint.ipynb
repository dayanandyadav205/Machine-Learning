```python
import numpy as np

def checkerboard(N):
    # Create an NxN array of zeros
    board = np.zeros((N, N), dtype=int)

    # Fill alternating positions with 1
    board[1::2, ::2] = 1   # Odd rows, even columns
    board[::2, 1::2] = 1   # Even rows, odd columns

    return board

# Example usage
N = 8
print(checkerboard(N))
```
