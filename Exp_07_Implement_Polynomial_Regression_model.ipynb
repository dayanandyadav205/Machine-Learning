```python
import pandas as pd
import numpy as np
import matplotlib.pyplot as plt
from sklearn.linear_model import LinearRegression
from sklearn.preprocessing import PolynomialFeatures

# Load dataset
df = pd.read_csv("Salary_data.csv")
X = df[['YearsExperience']]
y = df['Salary']

# --- Linear Regression ---
lin_reg = LinearRegression()
lin_reg.fit(X, y)

# --- Polynomial Regression (degree=3 for demonstration) ---
poly = PolynomialFeatures(degree=3)
X_poly = poly.fit_transform(X)
poly_reg = LinearRegression()
poly_reg.fit(X_poly, y)

# --- Visualization ---
plt.scatter(X, y, color='blue', label='Actual Data')

# Linear Regression line
plt.plot(X, lin_reg.predict(X), color='red', label='Linear Regression')

# Polynomial Regression curve
X_range = np.linspace(X.min(), X.max(), 100).reshape(-1,1)
plt.plot(X_range, poly_reg.predict(poly.transform(X_range)), color='green', label='Polynomial Regression (degree=3)')

plt.xlabel("Years of Experience")
plt.ylabel("Salary")
plt.title("Linear vs Polynomial Regression")
plt.legend()
plt.show()

- Blue scatter points → actual data.
- Red line → linear regression fit (straight line).
- Green curve → polynomial regression fit (curved line, degree=3).
- The polynomial curve usually fits closer to the data if the relationship is non-linear.
This comparison helps you decide whether a simple linear model is sufficient or if a polynomial model captures the trend better.
```
